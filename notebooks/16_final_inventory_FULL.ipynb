````python

# =============================================================================
# NB16_FULL — Consolidação, Inventário, Rastreabilidade e Fechamento do ramo _FULL
# Pipeline PPCOMP/DM — Dissertação de Mestrado
#
# Regra central:
# - Este notebook NÃO recalcula modelos, limiares, estados, episódios, escores,
#   custos, métricas ou figuras.
# - Este notebook apenas consolida, inventaria, valida e documenta os artefatos
#   já produzidos no downstream _FULL.
# =============================================================================

from __future__ import annotations

import os
import re
import json
import math
import hashlib
import warnings
from dataclasses import dataclass, asdict
from pathlib import Path
from datetime import datetime
from typing import Any, Dict, Iterable, List, Optional, Sequence, Tuple

import pandas as pd

warnings.filterwarnings("ignore")

# -----------------------------------------------------------------------------
# 1. Configuração de ambiente e parâmetros imutáveis
# -----------------------------------------------------------------------------

NB16_FULL_VERSION = "v1.4_nb06_expected_reports_ok"
EXECUTION_TS = datetime.now().isoformat(timespec="seconds")

try:
    from google.colab import drive  # type: ignore
    if not Path("/content/drive").exists() or not any(Path("/content/drive").iterdir()):
        drive.mount("/content/drive")
except Exception:
    # Execução fora do Colab: seguir usando caminhos locais, se existirem.
    pass

PROJECT_ROOT_CANDIDATES = [
    Path("/content/drive/MyDrive/Mestrado"),
    Path("/content/drive/My Drive/Mestrado"),
    Path.cwd(),
]
PROJECT_ROOT = next((p for p in PROJECT_ROOT_CANDIDATES if p.exists()), Path.cwd())

FULL_REPORTS_DIR_CANDIDATES = [
    PROJECT_ROOT / "04-reports" / "99_FULL_downstream",
    Path("/content/drive/MyDrive/Mestrado/04-reports/99_FULL_downstream"),
    Path("/content/drive/My Drive/Mestrado/04-reports/99_FULL_downstream"),
    Path.cwd() / "04-reports" / "99_FULL_downstream",
]
FULL_REPORTS_DIR = next((p for p in FULL_REPORTS_DIR_CANDIDATES if p.exists()), FULL_REPORTS_DIR_CANDIDATES[0])

STAGE16_DIR = FULL_REPORTS_DIR / "16_FULL_inventory"
AGGREGATE_DIR = STAGE16_DIR / "aggregate"
AGGREGATE_DIR.mkdir(parents=True, exist_ok=True)

FULL_MODEL_FACING_CANDIDATES = [
    PROJECT_ROOT / "02-datasets" / "99-full" / "03-model-facing" / "window_5min_series_allcells_model_facing_000000000000.parquet",
    PROJECT_ROOT / "02-datasets" / "99-full" / "03-model-facing" / "window_5min_series_allcells_model_facing_000000000000.parq",
    PROJECT_ROOT / "02-datasets" / "99-full" / "03-model-facing" / "window_5min_series_allcells_model_facing.parquet",
    Path("/content/drive/MyDrive/Mestrado/02-datasets/99-full/03-model-facing/window_5min_series_allcells_model_facing_000000000000.parquet"),
    Path("/content/drive/My Drive/Mestrado/02-datasets/99-full/03-model-facing/window_5min_series_allcells_model_facing_000000000000.parquet"),
]
FULL_MODEL_FACING_PARQUET = next((p for p in FULL_MODEL_FACING_CANDIDATES if p.exists()), FULL_MODEL_FACING_CANDIDATES[0])

FULL_CELLS = list("abcdefgh")
ACTIVE_CELLS = FULL_CELLS.copy()
SCENARIO_MAIN = "W5_K24_H12_P1_TRAIN_M2S"

CANONICAL_TEMPORAL_CORE = [
    "fail_rate",
    "n_events",
    "n_failed",
    "n_machines",
    "n_collections",
    "event_FAIL_count",
    "event_LOST_count",
    "lag_1",
    "lag_2",
    "lag_3",
    "rolling_mean_1h",
    "rolling_std_1h",
    "pct_change",
    "zscore_expanding",
]
assert len(CANONICAL_TEMPORAL_CORE) == 14

HASH_MAX_BYTES = 800 * 1024 * 1024  # 800 MB
EXCLUDE_DIR_PARTS = {".git", "__pycache__", ".ipynb_checkpoints", "backup", "backups", "_backup"}
INCLUDE_BACKUPS = False
EXCLUDE_NB16_OUTPUTS_FROM_SOURCE_INVENTORY = True

assert "99_FULL_downstream" in str(FULL_REPORTS_DIR), "NB16_FULL deve operar dentro de 99_FULL_downstream."
assert "03-features" not in str(STAGE16_DIR), "NB16_FULL não pode escrever em 03-features."

print("=" * 100)
print("NB16_FULL — Consolidação, Inventário, Rastreabilidade e Fechamento")
print("=" * 100)
print(f"Versão        : {NB16_FULL_VERSION}")
print(f"Execução      : {EXECUTION_TS}")
print(f"PROJECT_ROOT  : {PROJECT_ROOT}")
print(f"FULL_REPORTS  : {FULL_REPORTS_DIR}")
print(f"STAGE16_DIR   : {STAGE16_DIR}")
print(f"AGGREGATE_DIR : {AGGREGATE_DIR}")
print(f"Parquet FULL  : {FULL_MODEL_FACING_PARQUET}")
print(f"ACTIVE_CELLS  : {''.join(ACTIVE_CELLS)}")
print("=" * 100)

# -----------------------------------------------------------------------------
# 2. Funções utilitárias
# -----------------------------------------------------------------------------

def normalize_cell_id(value: Any) -> Optional[str]:
    if value is None:
        return None
    s = str(value).strip().lower()
    m = re.search(r"cell[_\- ]?([a-h])", s)
    if m:
        return m.group(1)
    if s in FULL_CELLS:
        return s
    return None


def sha256_file(path: Path, max_bytes: int = HASH_MAX_BYTES) -> Tuple[Optional[str], str]:
    try:
        if not path.exists() or not path.is_file():
            return None, "missing"
        size = path.stat().st_size
        if size > max_bytes:
            return None, f"skipped_gt_{max_bytes}_bytes"
        h = hashlib.sha256()
        with path.open("rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                h.update(chunk)
        return h.hexdigest(), "ok"
    except Exception as exc:
        return None, f"error:{type(exc).__name__}:{exc}"


def safe_read_json(path: Path) -> Optional[Any]:
    try:
        if not path.exists():
            return None
        with path.open("r", encoding="utf-8") as f:
            return json.load(f)
    except UnicodeDecodeError:
        try:
            with path.open("r", encoding="latin-1") as f:
                return json.load(f)
        except Exception:
            return None
    except Exception:
        return None


def safe_read_text(path: Path, max_chars: int = 20000) -> Optional[str]:
    try:
        if not path.exists():
            return None
        return path.read_text(encoding="utf-8", errors="replace")[:max_chars]
    except Exception:
        return None


def safe_read_csv(path: Path) -> Optional[pd.DataFrame]:
    try:
        if not path.exists():
            return None
        return pd.read_csv(path)
    except Exception:
        try:
            return pd.read_csv(path, sep=";")
        except Exception:
            return None


def safe_read_table(path: Path) -> Optional[pd.DataFrame]:
    suffix = path.suffix.lower()
    try:
        if not path.exists():
            return None
        if suffix == ".parquet":
            return pd.read_parquet(path)
        if suffix in {".csv", ".tsv"}:
            return pd.read_csv(path, sep="\t" if suffix == ".tsv" else ",")
        return None
    except Exception:
        return safe_read_csv(path)


def safe_read_parquet_shape(path: Path) -> Tuple[Optional[int], Optional[int], str]:
    try:
        if not path.exists():
            return None, None, "missing"
        df = pd.read_parquet(path)
        return int(df.shape[0]), int(df.shape[1]), "ok"
    except Exception as exc:
        return None, None, f"not_read:{type(exc).__name__}"


def flatten_json(obj: Any, prefix: str = "") -> Dict[str, Any]:
    out: Dict[str, Any] = {}
    if isinstance(obj, dict):
        for k, v in obj.items():
            new_prefix = f"{prefix}.{k}" if prefix else str(k)
            out.update(flatten_json(v, new_prefix))
    elif isinstance(obj, list):
        out[f"{prefix}.__len__"] = len(obj)
        for idx, item in enumerate(obj[:5]):
            out.update(flatten_json(item, f"{prefix}[{idx}]"))
    else:
        out[prefix] = obj
    return out


def as_scalar(value: Any) -> Any:
    if isinstance(value, (str, int, float, bool)) or value is None:
        return value
    try:
        if pd.isna(value):
            return None
    except Exception:
        pass
    return json.dumps(value, ensure_ascii=False, default=str)


def relpath(path: Path, base: Path = FULL_REPORTS_DIR) -> str:
    try:
        return str(path.relative_to(base)).replace("\\", "/")
    except Exception:
        try:
            return str(path.relative_to(PROJECT_ROOT)).replace("\\", "/")
        except Exception:
            return str(path).replace("\\", "/")


def artifact_kind(path: Path) -> str:
    suffix = path.suffix.lower()
    name = path.name.lower()
    if suffix in {".png", ".jpg", ".jpeg", ".pdf", ".svg", ".webp"}:
        return "figure_or_visual"
    if suffix in {".csv", ".tsv", ".xlsx", ".xls"}:
        return "table"
    if suffix == ".json":
        return "json_summary"
    if suffix in {".parquet", ".feather", ".parq"}:
        return "dataframe_snapshot"
    if suffix in {".txt", ".md"}:
        return "text_note"
    if suffix == ".ipynb":
        return "notebook"
    if suffix in {".py", ".sql", ".sh", ".ps1"}:
        return "code_or_script"
    if "fig" in name or "visual" in name:
        return "visual_related"
    return "other"


def is_under_path(path: Path, parent: Path) -> bool:
    """Retorna True se path estiver contido em parent, de forma robusta no Colab/local."""
    try:
        path.resolve().relative_to(parent.resolve())
        return True
    except Exception:
        return False


def is_nb16_generated_artifact(path: Path) -> bool:
    """Evita autorreferência direta ou indireta no inventário-fonte.

    Exclui:
    - a pasta oficial desta execução: 16_FULL_inventory/;
    - resíduos de execuções anteriores: _16_FULL_inventory/;
    - zips equivalentes, com ou sem sublinhado inicial;
    - qualquer arquivo 16_FULL_* salvo no downstream fora do STAGE16_DIR.
    """
    if not EXCLUDE_NB16_OUTPUTS_FROM_SOURCE_INVENTORY:
        return False

    # 1) Pasta oficial do NB16 atual.
    if is_under_path(path, STAGE16_DIR):
        return True

    try:
        rel_parts = [part.lower() for part in path.resolve().relative_to(FULL_REPORTS_DIR.resolve()).parts]
    except Exception:
        rel_parts = [part.lower() for part in path.parts]

    normalized_parts = {part.lstrip("_") for part in rel_parts}
    name_lower = path.name.lower()
    stem_norm = path.stem.lower().lstrip("_")

    # 2) Pastas antigas/renomeadas do próprio NB16.
    if "16_full_inventory" in normalized_parts:
        return True

    # 3) Zips antigos/renomeados do próprio NB16.
    if path.suffix.lower() == ".zip" and stem_norm in {"16_full_inventory", "16_full_inventory_final"}:
        return True

    # 4) Qualquer artefato 16_FULL_* perdido no downstream deve ser tratado como output NB16,
    #    não como artefato-fonte de NB04–NB15.
    if name_lower.startswith("16_full_") and (FULL_REPORTS_DIR in path.parents or is_under_path(path, FULL_REPORTS_DIR)):
        return True

    return False


def is_excluded_path(path: Path) -> bool:
    parts_lower = {part.lower() for part in path.parts}

    if any(part in EXCLUDE_DIR_PARTS for part in parts_lower):
        return True

    if not INCLUDE_BACKUPS and any("backup" in part.lower() for part in path.parts):
        return True

    # O inventário principal deve representar as entradas do fechamento,
    # não os artefatos gerados pelo próprio NB16_FULL.
    if is_nb16_generated_artifact(path):
        return True

    return False


def infer_stage_dir_from_path(path: Path) -> str:
    r = relpath(path)
    first = r.split("/")[0] if r else ""
    return first


def infer_nb_from_stage(stage_dir: str) -> Optional[str]:
    s = stage_dir.lower()
    if s.startswith("13a_full"):
        return "13a"
    m = re.match(r"^(\d{2})_full", s)
    if m:
        return m.group(1)
    return None


def infer_cell_from_path(path: Path) -> Optional[str]:
    for part in path.parts:
        cid = normalize_cell_id(part)
        if cid:
            return cid
    cid = normalize_cell_id(path.name)
    if cid:
        return cid
    return None


def first_col(df: pd.DataFrame, candidates: Sequence[str]) -> Optional[str]:
    if df is None or df.empty:
        return None
    lower_map = {c.lower(): c for c in df.columns}
    for cand in candidates:
        if cand.lower() in lower_map:
            return lower_map[cand.lower()]
    for cand in candidates:
        for c in df.columns:
            if cand.lower() in c.lower():
                return c
    return None


def add_result(rows: List[Dict[str, Any]], source: str, indicator: str, value: Any,
               interpretation: str = "", cell_id: str = "", stage: str = "") -> None:
    rows.append({
        "stage": stage,
        "cell_id": cell_id,
        "source": source,
        "indicator": indicator,
        "value": as_scalar(value),
        "interpretation": interpretation,
    })


def add_check(rows: List[Dict[str, Any]], check: str, status: str, evidence: str = "", severity: str = "info") -> None:
    rows.append({
        "check": check,
        "status": status,
        "severity": severity,
        "evidence": evidence,
    })


def glob_count(base: Path, pattern: str) -> Tuple[int, str]:
    if not base.exists():
        return 0, ""
    matches = sorted(base.glob(pattern))
    example = "; ".join(relpath(p) for p in matches[:5])
    return len(matches), example


def expected_pattern_status(found_count: int, required_stage: bool) -> str:
    if found_count > 0:
        return "OK"
    return "MISSING_REQUIRED" if required_stage else "MISSING_OPTIONAL"


def save_csv(df: pd.DataFrame, filename: str) -> Path:
    path = AGGREGATE_DIR / filename
    path.parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(path, index=False, encoding="utf-8")
    return path


def save_json(obj: Any, filename: str) -> Path:
    path = AGGREGATE_DIR / filename
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(json.dumps(obj, ensure_ascii=False, indent=2, default=str), encoding="utf-8")
    return path


def save_md(text: str, filename: str) -> Path:
    path = AGGREGATE_DIR / filename
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(text, encoding="utf-8")
    return path

# -----------------------------------------------------------------------------
# 3. Especificação das etapas _FULL esperadas
# -----------------------------------------------------------------------------

@dataclass
class StageSpec:
    nb: str
    stage_dir: str
    title: str
    role: str
    required: bool
    structure: str
    expected_aggregate_patterns: List[str]
    expected_cell_patterns: List[str]
    notes: str = ""

STAGE_SPECS: List[StageSpec] = [
    StageSpec(
        "04",
        "04_FULL_episodes",
        "Episódios por célula",
        "Detectar episódios críticos e registrar modelabilidade inicial por célula.",
        True,
        "cell_dirs",
        [
            "04_FULL_detect_episodes_summary.json",
            "04_FULL_episode_summary_by_cell.csv",
            "04_FULL_modelability_gate_by_cell.csv",
            "04_FULL_thresholds_by_cell.csv",
            "04_FULL_artifact_manifest_sha256.csv",
        ],
        ["*04_FULL*cell_{cell}*", "*episodes*cell_{cell}*"],
        "Início efetivo do downstream _FULL; padrões agregados alinhados aos nomes reais.",
    ),
    StageSpec(
        "05",
        "05_FULL_features",
        "Features causais por célula",
        "Construir atributos temporais causais preservando temporal_core.",
        True,
        "cell_dirs",
        [
            "05_FULL_feature_engineering_summary.json",
            "05_FULL_feature_summary_by_cell.csv",
            "05_FULL_feature_sets.json",
            "05_FULL_artifact_manifest_sha256.csv",
        ],
        ["*05_FULL*cell_{cell}*", "*features*cell_{cell}*"],
        "Padrões agregados alinhados aos nomes reais do NB05_FULL.",
    ),
    StageSpec(
        "06",
        "06_FULL_labeling_states",
        "Rotulagem e estados",
        "Gerar estados NORMAL/BEFORE/DURING/AFTER e alvo supervisionado por célula/cenário.",
        True,
        "nb06_exception",
        [
            "06_FULL_labeling_states_summary.json",
            "06_FULL_horizon_sensitivity_summary.json",
            "06_FULL_scenario_summary_table.csv",
            "06_FULL_primary_diagnostic_flags.csv",
            "06_FULL_artifact_manifest_sha256.csv",
        ],
        ["*cell_{cell}*"],
        "Exceção controlada: usa subpastas features/, reports/ e aggregate/. Padrões agregados principais são buscados em reports/.",
    ),
    StageSpec(
        "10",
        "10_FULL_threshold_diagnostics",
        "Diagnósticos de cenário",
        "Consolidar cenários, limiares e decisões por célula.",
        True,
        "cell_dirs",
        [
            "10_FULL_diagnostics_summary.json",
            "10_FULL_scenario_decisions_by_cell.csv",
            "10_FULL_thresholds_summary_by_cell.csv",
            "10_FULL_episode_summary_by_cell.csv",
            "10_FULL_artifact_manifest_sha256.csv",
        ],
        ["*10_FULL*cell_{cell}*", "*scenario*cell_{cell}*"],
        "Padrões agregados alinhados aos nomes reais do NB10_FULL.",
    ),
    StageSpec(
        "11",
        "11_FULL_model_baselines",
        "Modelos e baselines",
        "Comparar modelos tabulares, regras temporais e baseline EWMA por célula.",
        True,
        "cell_dirs",
        [
            "11_FULL_nb11_summary.json",
            "11_FULL_winner_model.json",
            "11_FULL_metrics_summary_by_cell.csv",
            "11_FULL_metrics_aggregate.csv",
            "11_FULL_artifact_manifest_sha256.csv",
        ],
        ["*11_FULL*cell_{cell}*", "*score*cell_{cell}*"],
        "Padrões agregados alinhados aos nomes reais do NB11_FULL.",
    ),
    StageSpec(
        "12",
        "12_FULL_sensitivity",
        "Sensibilidade e robustez",
        "Avaliar robustez compacta por célula e preservar decisões centrais.",
        True,
        "cell_dirs",
        [
            "12_FULL_nb12_summary.json",
            "12_FULL_sensitivity_summary_by_cell.csv",
            "12_FULL_sensitivity_aggregate.csv",
            "12_FULL_winner_context_by_cell.csv",
            "12_FULL_artifact_manifest_sha256.csv",
        ],
        ["*12_FULL*cell_{cell}*", "*sensitivity*cell_{cell}*"],
        "Padrões agregados alinhados aos nomes reais do NB12_FULL.",
    ),
    StageSpec(
        "13",
        "13_FULL_lstm",
        "LSTM complementar",
        "Registrar comparação sequencial complementar, sem deslocar o núcleo tabular.",
        True,
        "cell_dirs",
        [
            "13_FULL_lstm_summary.json",
            "13_FULL_lstm_fixed_results_by_cell.csv",
            "13_FULL_lstm_tscv_results_by_cell.csv",
            "13_FULL_lstm_delta_uncertainty_by_cell.csv",
            "13_FULL_lstm_decision_to_nb14_by_cell.csv",
            "13_FULL_lstm_skips_by_cell.csv",
            "13_FULL_pilot_validation_summary.csv",
        ],
        ["*13_FULL*cell_{cell}*", "*lstm*cell_{cell}*"],
        "Artefatos agregados possuem nomes específicos do NB13_FULL; não usar padrões genéricos.",
    ),
    StageSpec(
        "13a",
        "13a_FULL_lstm_tuning",
        "Tuning LSTM complementar",
        "Registrar ajuste compacto complementar quando executado.",
        False,
        "optional_cell_dirs",
        [
            "13a_FULL_lstm_tuning_summary.json",
            "13a_FULL_lstm_tuning_decision_to_nb14_by_cell.csv",
            "13a_FULL_lstm_tuning_delta_vs_NB11_FULL_by_cell.csv",
            "13a_FULL_lstm_tuning_delta_vs_NB13_FULL_by_cell.csv",
            "13a_FULL_lstm_tuning_fixed_results_by_cell.csv",
            "13a_FULL_lstm_tuning_issues_allcells.csv",
            "13a_FULL_pilot_validation_summary.csv",
        ],
        ["*13a_FULL*cell_{cell}*", "*tuning*cell_{cell}*"],
        "Opcional e complementar; ausência não bloqueia fechamento.",
    ),
    StageSpec(
        "14",
        "14_FULL_decision_analysis",
        "Análise decisória",
        "Traduzir escores em custo, tau, lead time e antecipabilidade por célula.",
        True,
        "cell_dirs",
        [
            "14_FULL_scenario_summary_by_cell.csv",
            "14_FULL_cost_summary_by_cell.csv",
            "14_FULL_optimal_tau_by_cost_allcells.csv",
            "14_FULL_cost_curve_allcells.csv",
            "14_FULL_nb14_summary.json",
        ],
        ["*14_FULL*cell_{cell}*", "*cost*cell_{cell}*", "*score*cell_{cell}*"],
        "Consumido pelo NB15_FULL.",
    ),
    StageSpec(
        "15",
        "15_FULL_visualization",
        "Visualização final",
        "Gerar figuras finais por célula e painéis agregados.",
        True,
        "cell_dirs",
        [
            "15_FULL_visual_manifest_allcells.csv",
            "15_FULL_calibration_diagnostics_by_cell.csv",
            "15_FULL_delta_vs_canonical.csv",
            "15_FULL_artifact_manifest_sha256.csv",
            "15_FULL_pilot_validation_summary.csv",
            "15_FULL_nb15_summary.json",
        ],
        ["*15_FULL*cell_{cell}*", "figures/*cell_{cell}*"],
        "Entrada imediata do NB16_FULL.",
    ),
]
STAGE_BY_DIR = {spec.stage_dir: spec for spec in STAGE_SPECS}

# -----------------------------------------------------------------------------
# 4. Inventário de artefatos
# -----------------------------------------------------------------------------

artifact_paths: List[Path] = []
if FULL_REPORTS_DIR.exists():
    for p in FULL_REPORTS_DIR.rglob("*"):
        if is_excluded_path(p):
            continue
        if p.is_file():
            artifact_paths.append(p)

# Inclui proveniência do Parquet model-facing se existir.
if FULL_MODEL_FACING_PARQUET.exists():
    artifact_paths.append(FULL_MODEL_FACING_PARQUET)

artifact_paths = sorted(set(artifact_paths), key=lambda x: str(x).lower())

inventory_rows: List[Dict[str, Any]] = []
manifest_rows: List[Dict[str, Any]] = []

for p in artifact_paths:
    stage_dir = infer_stage_dir_from_path(p) if FULL_REPORTS_DIR in p.parents or p == FULL_REPORTS_DIR else "FULL_MODEL_FACING_INPUT"
    if p == FULL_MODEL_FACING_PARQUET:
        stage_dir = "FULL_MODEL_FACING_INPUT"
        nb = "99_C"
        layer = "upstream_full_materialization"
        dsr_cycle = "full_pre_downstream"
    else:
        nb = infer_nb_from_stage(stage_dir)
        layer = "downstream_full" if nb else "fora_do_escopo_full"
        dsr_cycle = "full" if nb else "unknown"
    stat = p.stat()
    h, hs = sha256_file(p)
    row = {
        "nb": nb,
        "stage_dir": stage_dir,
        "layer": layer,
        "dsr_cycle": dsr_cycle,
        "cell_id": infer_cell_from_path(p) or "",
        "artifact_kind": artifact_kind(p),
        "filename": p.name,
        "relative_path": relpath(p),
        "absolute_path": str(p),
        "suffix": p.suffix.lower(),
        "size_bytes": int(stat.st_size),
        "size_mb": round(stat.st_size / (1024 * 1024), 4),
        "modified_at": datetime.fromtimestamp(stat.st_mtime).isoformat(timespec="seconds"),
        "sha256": h,
        "sha256_status": hs,
    }
    inventory_rows.append(row)
    manifest_rows.append({k: row[k] for k in ["nb", "stage_dir", "cell_id", "filename", "relative_path", "size_bytes", "modified_at", "sha256", "sha256_status"]})

inventory_df = pd.DataFrame(inventory_rows)
manifest_df = pd.DataFrame(manifest_rows)

# Sanitização final defensiva: garante que resíduos do próprio NB16 não entrem
# no inventário-fonte nem no manifesto-fonte, mesmo se estiverem fora da pasta
# oficial atual, por exemplo em _16_FULL_inventory/ ou _16_FULL_inventory.zip.
def is_nb16_residual_row(row: pd.Series) -> bool:
    rel = str(row.get("relative_path", "")).lower()
    fname = str(row.get("filename", "")).lower()
    stage = str(row.get("stage_dir", "")).lower()
    rel_norm = rel.replace("\\", "/")
    parts_norm = {part.lstrip("_") for part in rel_norm.split("/") if part}
    stage_norm = stage.lstrip("_")
    stem_norm = Path(fname).stem.lower().lstrip("_")

    if "16_full_inventory" in parts_norm or stage_norm == "16_full_inventory":
        return True
    if fname.endswith(".zip") and stem_norm in {"16_full_inventory", "16_full_inventory_final"}:
        return True
    if fname.startswith("16_full_"):
        return True
    return False

if not inventory_df.empty:
    residual_mask = inventory_df.apply(is_nb16_residual_row, axis=1)
    nb16_residuals_removed = int(residual_mask.sum())
    if nb16_residuals_removed:
        print(f"Resíduos NB16 removidos do inventário-fonte: {nb16_residuals_removed}")
        inventory_df = inventory_df.loc[~residual_mask].reset_index(drop=True)
        manifest_df = manifest_df.loc[~residual_mask].reset_index(drop=True)
else:
    nb16_residuals_removed = 0

print(f"Artefatos inventariados: {len(inventory_df)}")
if not inventory_df.empty:
    group_cols = ["stage_dir", "nb", "artifact_kind"]
    print(inventory_df.groupby(group_cols).size().reset_index(name="n").to_string(index=False))

# -----------------------------------------------------------------------------
# 5. Rastreabilidade por etapa, cobertura por célula e checks esperados
# -----------------------------------------------------------------------------

trace_rows: List[Dict[str, Any]] = []
coverage_rows: List[Dict[str, Any]] = []
expected_rows: List[Dict[str, Any]] = []

for spec in STAGE_SPECS:
    stage_path = FULL_REPORTS_DIR / spec.stage_dir
    aggregate_path = stage_path / "aggregate"
    dir_exists = stage_path.exists() and stage_path.is_dir()
    aggregate_exists = aggregate_path.exists() and aggregate_path.is_dir()
    stage_files = [p for p in artifact_paths if stage_path in p.parents] if dir_exists else []
    stage_file_count = len(stage_files)
    stage_size_mb = round(sum(p.stat().st_size for p in stage_files if p.exists()) / (1024 * 1024), 4)

    # Verificação de padrões agregados.
    # Regra: etapas convencionais procuram em aggregate/; NB06_FULL é exceção
    # controlada e mantém summaries/manifests principais em reports/.
    # A versão v1.3 buscava esses padrões na raiz da etapa, o que gerava
    # MISSING_REQUIRED falso em 06_FULL_labeling_states, embora os arquivos
    # existissem em 06_FULL_labeling_states/reports/.
    expected_search_base = (stage_path / "reports") if spec.structure == "nb06_exception" else aggregate_path
    found_agg_patterns = 0
    missing_agg_patterns = []
    for pat in spec.expected_aggregate_patterns:
        cnt, ex = glob_count(expected_search_base, pat)
        if cnt > 0:
            found_agg_patterns += 1
        else:
            missing_agg_patterns.append(pat)
        expected_rows.append({
            "nb": spec.nb,
            "stage_dir": spec.stage_dir,
            "scope": "aggregate" if spec.structure != "nb06_exception" else "stage_dir",
            "search_base": relpath(expected_search_base),
            "pattern": pat,
            "found_count": cnt,
            "examples": ex,
            "required_stage": spec.required,
            "status": expected_pattern_status(cnt, spec.required),
        })

    # Cobertura por célula.
    cells_with_files = set()
    for p in stage_files:
        cid = infer_cell_from_path(p)
        if cid in FULL_CELLS:
            cells_with_files.add(cid)

    if spec.structure in {"cell_dirs", "optional_cell_dirs"}:
        for cid in ACTIVE_CELLS:
            cell_dir = stage_path / f"cell_{cid}"
            cell_files = [p for p in stage_files if cell_dir in p.parents] if cell_dir.exists() else []
            by_name_files = [p for p in stage_files if f"cell_{cid}" in p.name.lower()]
            total_cell_files = len(set(cell_files + by_name_files))
            cell_status = "OK" if total_cell_files > 0 else ("SKIP_OPTIONAL" if not spec.required else "WARN")
            coverage_rows.append({
                "nb": spec.nb,
                "stage_dir": spec.stage_dir,
                "cell_id": cid,
                "cell_dir_exists": cell_dir.exists(),
                "artifact_count": total_cell_files,
                "status": cell_status,
            })
            if total_cell_files > 0:
                cells_with_files.add(cid)

            for pat in spec.expected_cell_patterns:
                pattern = pat.format(cell=cid)
                cnt1, ex1 = glob_count(cell_dir, pattern) if cell_dir.exists() else (0, "")
                # fallback: procurar no stage inteiro pelo nome da célula.
                cnt2, ex2 = glob_count(stage_path, f"**/{pattern}") if dir_exists else (0, "")
                cnt = max(cnt1, cnt2)
                ex = ex1 or ex2
                expected_rows.append({
                    "nb": spec.nb,
                    "stage_dir": spec.stage_dir,
                    "scope": f"cell_{cid}",
                    "pattern": pattern,
                    "found_count": cnt,
                    "examples": ex,
                    "required_stage": spec.required,
                    "status": expected_pattern_status(cnt, spec.required),
                })
    elif spec.structure == "nb06_exception":
        for sub in ["features", "reports", "aggregate"]:
            sub_path = stage_path / sub
            cnt = len([p for p in sub_path.rglob("*") if p.is_file()]) if sub_path.exists() else 0
            coverage_rows.append({
                "nb": spec.nb,
                "stage_dir": spec.stage_dir,
                "cell_id": sub,
                "cell_dir_exists": sub_path.exists(),
                "artifact_count": cnt,
                "status": "OK" if cnt > 0 or sub == "aggregate" and aggregate_exists else "WARN",
            })
        # Tenta inferir células por nome de arquivo.
        for cid in ACTIVE_CELLS:
            total_cell_files = len([p for p in stage_files if f"cell_{cid}" in p.name.lower() or f"_{cid}_" in p.name.lower()])
            coverage_rows.append({
                "nb": spec.nb,
                "stage_dir": spec.stage_dir,
                "cell_id": cid,
                "cell_dir_exists": False,
                "artifact_count": total_cell_files,
                "status": "OK" if total_cell_files > 0 else "WARN",
            })
            if total_cell_files > 0:
                cells_with_files.add(cid)

    if not dir_exists and not spec.required:
        status = "SKIP_OPTIONAL"
    elif not dir_exists and spec.required:
        status = "FAIL"
    elif spec.structure == "nb06_exception":
        status = "OK" if aggregate_exists and stage_file_count > 0 else "WARN"
    elif spec.required and len(cells_with_files.intersection(ACTIVE_CELLS)) < len(ACTIVE_CELLS):
        status = "WARN"
    elif spec.required and found_agg_patterns == 0:
        status = "WARN"
    else:
        status = "OK"

    trace_rows.append({
        "nb": spec.nb,
        "stage_dir": spec.stage_dir,
        "title": spec.title,
        "pipeline_role": spec.role,
        "required": spec.required,
        "structure": spec.structure,
        "dir_exists": dir_exists,
        "aggregate_exists": aggregate_exists,
        "artifact_count": stage_file_count,
        "size_mb": stage_size_mb,
        "cells_with_artifacts": "".join(sorted(cells_with_files.intersection(ACTIVE_CELLS))),
        "cells_with_artifacts_count": len(cells_with_files.intersection(ACTIVE_CELLS)),
        "expected_aggregate_patterns": len(spec.expected_aggregate_patterns),
        "expected_aggregate_patterns_found": found_agg_patterns,
        "missing_aggregate_patterns": "; ".join(missing_agg_patterns),
        "status": status,
        "notes": spec.notes,
    })

traceability_df = pd.DataFrame(trace_rows)
coverage_df = pd.DataFrame(coverage_rows)
expected_df = pd.DataFrame(expected_rows)

print("\nMatriz de rastreabilidade — resumo:")
print(traceability_df[["nb", "stage_dir", "artifact_count", "cells_with_artifacts_count", "status"]].to_string(index=False))

# -----------------------------------------------------------------------------
# 6. Extração documental de resultados-chave já consolidados
# -----------------------------------------------------------------------------

key_results: List[Dict[str, Any]] = []
consistency_rows: List[Dict[str, Any]] = []

# Checks estruturais centrais.
add_check(consistency_rows, "Diretório FULL downstream", "OK" if "99_FULL_downstream" in str(FULL_REPORTS_DIR) else "FAIL", str(FULL_REPORTS_DIR), "critical" if "99_FULL_downstream" not in str(FULL_REPORTS_DIR) else "info")
add_check(consistency_rows, "Saída fora de 03-features", "OK" if "03-features" not in str(STAGE16_DIR) else "FAIL", str(STAGE16_DIR), "critical" if "03-features" in str(STAGE16_DIR) else "info")
add_check(consistency_rows, "Parquet model-facing FULL localizado", "OK" if FULL_MODEL_FACING_PARQUET.exists() else "WARN", str(FULL_MODEL_FACING_PARQUET), "warning" if not FULL_MODEL_FACING_PARQUET.exists() else "info")
for nb_missing in ["07", "08", "09"]:
    exists_unexpected = any(
        (FULL_REPORTS_DIR / f"{nb_missing}_FULL{x}").exists()
        for x in ["", "_", "_modeling"]
    )
    add_check(
        consistency_rows,
        f"Ausência esperada de NB{nb_missing}_FULL",
        "WARN" if exists_unexpected else "OK",
        f"NB{nb_missing}_FULL não é esperado no ramo _FULL.",
        "warning" if exists_unexpected else "info",
    )

# Possível risco de nomes canônicos sem _FULL dentro do downstream.
canonical_name_risks = []
if not inventory_df.empty:
    for _, row in inventory_df.iterrows():
        fn = str(row.get("filename", ""))
        stage_dir = str(row.get("stage_dir", ""))
        if stage_dir.startswith("FULL_MODEL"):
            continue
        if re.match(r"^(0[4-6]|1[0-6]|13a)_", fn) and "_FULL" not in fn:
            canonical_name_risks.append(str(row.get("relative_path", fn)))
add_check(
    consistency_rows,
    "Nomes canônicos sem _FULL dentro do downstream",
    "OK" if not canonical_name_risks else "WARN",
    "; ".join(canonical_name_risks[:20]) if canonical_name_risks else "nenhum risco identificado",
    "warning" if canonical_name_risks else "info",
)

# Checks por etapa.
for _, row in traceability_df.iterrows():
    sev = "critical" if row["status"] == "FAIL" else ("warning" if row["status"] == "WARN" else "info")
    add_check(consistency_rows, f"Etapa {row['nb']} — {row['stage_dir']}", row["status"], f"artefatos={row['artifact_count']}; células={row['cells_with_artifacts']}", sev)

# Proveveniência do Parquet model-facing.
if FULL_MODEL_FACING_PARQUET.exists():
    n_rows, n_cols, shape_status = safe_read_parquet_shape(FULL_MODEL_FACING_PARQUET)
    add_result(key_results, "FULL_MODEL_FACING_PARQUET", "exists", True, "Entrada selada do ramo _FULL localizada.", stage="99_C")
    add_result(key_results, "FULL_MODEL_FACING_PARQUET", "shape", {"rows": n_rows, "cols": n_cols, "status": shape_status}, "Forma lida apenas para proveniência; não há recálculo.", stage="99_C")

# NB14 aggregate — principal insumo decisório do NB15/NB16.
agg14 = FULL_REPORTS_DIR / "14_FULL_decision_analysis" / "aggregate"
nb14_files = {
    "scenario_summary_by_cell": agg14 / "14_FULL_scenario_summary_by_cell.csv",
    "cost_summary_by_cell": agg14 / "14_FULL_cost_summary_by_cell.csv",
    "optimal_tau_allcells": agg14 / "14_FULL_optimal_tau_by_cost_allcells.csv",
    "cost_curve_allcells": agg14 / "14_FULL_cost_curve_allcells.csv",
    "pilot_validation_summary": agg14 / "14_FULL_pilot_validation_summary.csv",
    "nb14_summary": agg14 / "14_FULL_nb14_summary.json",
}
for label, path in nb14_files.items():
    add_result(key_results, path.name, "exists", path.exists(), f"Artefato NB14_FULL aggregate: {label}.", stage="14")

df14_scenario = safe_read_csv(nb14_files["scenario_summary_by_cell"])
if df14_scenario is not None:
    add_result(key_results, nb14_files["scenario_summary_by_cell"].name, "rows", len(df14_scenario), "Número de linhas do resumo por célula.", stage="14")
    c_cell = first_col(df14_scenario, ["cell_id", "cell"])
    c_scenario = first_col(df14_scenario, ["scenario_label", "scenario"])
    c_score = first_col(df14_scenario, ["score_source", "source", "score_family"])
    if c_cell:
        add_result(key_results, nb14_files["scenario_summary_by_cell"].name, "cells", sorted(map(str, df14_scenario[c_cell].dropna().unique())), "Células representadas no resumo NB14_FULL.", stage="14")
    if c_scenario:
        add_result(key_results, nb14_files["scenario_summary_by_cell"].name, "scenario_counts", df14_scenario[c_scenario].value_counts(dropna=False).to_dict(), "Distribuição de cenários no NB14_FULL.", stage="14")
    if c_score:
        add_result(key_results, nb14_files["scenario_summary_by_cell"].name, "score_source_counts", df14_scenario[c_score].value_counts(dropna=False).to_dict(), "Fontes de escore usadas pelo NB14_FULL.", stage="14")

# NB15 aggregate — última etapa anterior.
agg15 = FULL_REPORTS_DIR / "15_FULL_visualization" / "aggregate"
nb15_files = {
    "visual_manifest_allcells": agg15 / "15_FULL_visual_manifest_allcells.csv",
    "calibration_diagnostics_by_cell": agg15 / "15_FULL_calibration_diagnostics_by_cell.csv",
    "delta_vs_canonical": agg15 / "15_FULL_delta_vs_canonical.csv",
    "artifact_manifest_sha256": agg15 / "15_FULL_artifact_manifest_sha256.csv",
    "pilot_validation_summary": agg15 / "15_FULL_pilot_validation_summary.csv",
    "nb15_summary": agg15 / "15_FULL_nb15_summary.json",
}
for label, path in nb15_files.items():
    add_result(key_results, path.name, "exists", path.exists(), f"Artefato NB15_FULL aggregate: {label}.", stage="15")

df15_manifest = safe_read_csv(nb15_files["visual_manifest_allcells"])
if df15_manifest is not None:
    add_result(key_results, nb15_files["visual_manifest_allcells"].name, "rows", len(df15_manifest), "Quantidade de registros do manifesto visual.", stage="15")
    c_cell = first_col(df15_manifest, ["cell_id", "cell"])
    c_fig = first_col(df15_manifest, ["figure_type", "fig_type"])
    if c_cell:
        add_result(key_results, nb15_files["visual_manifest_allcells"].name, "cells", sorted(map(str, df15_manifest[c_cell].dropna().unique())), "Células representadas no manifesto visual.", stage="15")
    if c_fig:
        add_result(key_results, nb15_files["visual_manifest_allcells"].name, "figure_type_counts", df15_manifest[c_fig].value_counts(dropna=False).to_dict(), "Tipos de figura gerados pelo NB15_FULL.", stage="15")

nb15_summary = safe_read_json(nb15_files["nb15_summary"])
if nb15_summary is not None:
    flat = flatten_json(nb15_summary)
    for fragments, label in [
        (["notebook"], "notebook"),
        (["version"], "version"),
        (["cells", "completed"], "cells_completed"),
        (["cells", "requested"], "cells_requested"),
        (["visual", "manifest"], "visual_manifest_path"),
        (["pilot"], "pilot_validation"),
    ]:
        candidates = [k for k in flat if all(f in k.lower() for f in fragments)]
        if candidates:
            add_result(key_results, nb15_files["nb15_summary"].name, label, flat[candidates[0]], "Valor extraído do summary NB15_FULL.", stage="15")

# Resumo dos status dos pilotos agregados, se existirem.
for p in [nb14_files["pilot_validation_summary"], nb15_files["pilot_validation_summary"]]:
    dfp = safe_read_csv(p)
    if dfp is not None:
        status_col = first_col(dfp, ["status", "result", "validated_status"])
        if status_col:
            add_result(key_results, p.name, "status_counts", dfp[status_col].value_counts(dropna=False).to_dict(), "Distribuição de status do pilot_validation_summary.", stage=infer_nb_from_stage(infer_stage_dir_from_path(p)) or "")

key_results_df = pd.DataFrame(key_results)
consistency_df = pd.DataFrame(consistency_rows)

# -----------------------------------------------------------------------------
# 7. Registro estruturado de decisões e limitações do ramo _FULL
# -----------------------------------------------------------------------------
#
# Esta seção espelha o papel documental do decision_log_df/limitations_df do NB16
# canônico.  O objetivo é transformar decisões governantes do ramo _FULL em
# artefatos citáveis, com evidências verificáveis no diretório downstream.
#
# Importante: as evidências abaixo são artefatos já produzidos por NB04_FULL–NB15_FULL.
# O NB16_FULL apenas verifica sua existência; não recalcula nem reabre resultados.

def evidence_lookup(patterns: Sequence[str]) -> Tuple[str, List[str], List[str]]:
    """Verifica se cada artefato/padrão citado possui ao menos uma evidência existente.

    Regras:
    - padrão relativo: procurado sob FULL_REPORTS_DIR;
    - caminho absoluto: verificado diretamente;
    - glob com *, ?, [] ou **: aceito;
    - status:
        existe  -> todos os padrões citados possuem evidência;
        parcial -> ao menos um padrão existe, mas há ausências;
        ausente -> nenhum padrão possui evidência;
        sem_artefato_definido -> lista vazia.
    """
    found: List[str] = []
    missing: List[str] = []

    for raw in patterns:
        pat = str(raw).strip()
        if not pat:
            continue

        matches: List[Path] = []

        # Caminho absoluto.
        p_abs = Path(pat)
        if p_abs.is_absolute():
            if p_abs.exists():
                matches = [p_abs]
        else:
            # Caminho relativo exato sob FULL_REPORTS_DIR.
            p_rel = FULL_REPORTS_DIR / pat
            if p_rel.exists():
                matches = [p_rel]
            else:
                # Glob relativo sob FULL_REPORTS_DIR.
                try:
                    matches = sorted([p for p in FULL_REPORTS_DIR.glob(pat) if p.exists()])
                except Exception:
                    matches = []

        # Manter apenas arquivos quando possível; diretórios também contam como
        # evidência apenas se o padrão explicitamente apontou para diretório.
        file_matches = [m for m in matches if m.is_file()]
        usable_matches = file_matches if file_matches else matches

        if usable_matches:
            found.extend(relpath(m) for m in usable_matches[:10])
        else:
            missing.append(pat)

    if not patterns:
        status = "sem_artefato_definido"
    elif found and not missing:
        status = "existe"
    elif found and missing:
        status = "parcial"
    else:
        status = "ausente"

    # Remover duplicatas preservando ordem.
    found_unique = list(dict.fromkeys(found))
    missing_unique = list(dict.fromkeys(missing))
    return status, found_unique, missing_unique


def build_evidence_df(specs: List[Dict[str, Any]]) -> pd.DataFrame:
    rows: List[Dict[str, Any]] = []
    for spec in specs:
        patterns = list(spec.get("evidence_patterns", []))
        status, found, missing = evidence_lookup(patterns)
        rows.append({
            "decision_id": spec.get("decision_id", ""),
            "limitation_id": spec.get("limitation_id", ""),
            "decision": spec.get("decision", ""),
            "limitation": spec.get("limitation", ""),
            "basis": spec.get("basis", ""),
            "final_position": spec.get("final_position", ""),
            "evidence_artifacts": "; ".join(patterns),
            "evidence_status": status,
            "evidence_found": "; ".join(found),
            "evidence_missing": "; ".join(missing),
        })
    return pd.DataFrame(rows)


decision_specs: List[Dict[str, Any]] = [
    {
        "decision_id": "DF01",
        "decision": "TRAIN_M2S é o cenário governante obrigatório no ramo _FULL; limiar μ+2σ global retrospectivo não é usado como base preditiva principal.",
        "basis": "O _FULL preserva a decisão metodológica de estimar criticidade com estatísticas do trecho de treino, reduzindo vazamento temporal e mantendo comparabilidade com o canônico.",
        "final_position": "Manter W5_K24_H12_P1_TRAIN_M2S como cenário principal; cenários globais/descritivos permanecem apenas como sensibilidade ou diagnóstico.",
        "evidence_patterns": [
            "04_FULL_episodes/aggregate/04_FULL_thresholds_by_cell.csv",
            "04_FULL_episodes/aggregate/04_FULL_episodes_TRAIN_M2S_active_cells.parquet",
            "10_FULL_threshold_diagnostics/aggregate/10_FULL_scenario_decisions_by_cell.csv",
        ],
    },
    {
        "decision_id": "DF02",
        "decision": "O split treino/teste é herdado dos artefatos upstream por célula; nenhum notebook downstream deve re-derivar train_cutoff_idx por fração local.",
        "basis": "O corte herdado evita deslocamento de fronteira temporal e contradição interna entre NB04/NB06/NB10+.",
        "final_position": "Preservar o corte materializado nos artefatos upstream; NB16_FULL apenas inventaria e registra a evidência, sem recomputar split.",
        "evidence_patterns": [
            "04_FULL_episodes/aggregate/04_FULL_detect_episodes_summary.json",
            "04_FULL_episodes/aggregate/04_FULL_modelability_gate_by_cell.csv",
            "11_FULL_model_baselines/aggregate/11_FULL_model_input_features_schema.json",
        ],
    },
    {
        "decision_id": "DF03",
        "decision": "A LSTM só poderia substituir o núcleo tabular sob gate Δ/SE ≥ 2,0; nenhuma célula promoveu LSTM como fonte soberana.",
        "basis": "NB13_FULL e NB13a_FULL foram mantidos como comparação/sensibilidade sequencial, mas a decisão soberana permanece conservadora.",
        "final_position": "LSTM complementar; NB11_FULL permanece como núcleo decisório primário.",
        "evidence_patterns": [
            "13_FULL_lstm/aggregate/13_FULL_lstm_delta_uncertainty_by_cell.csv",
            "13_FULL_lstm/aggregate/13_FULL_lstm_decision_to_nb14_by_cell.csv",
            "13a_FULL_lstm_tuning/aggregate/13a_FULL_lstm_tuning_decision_to_nb14_by_cell.csv",
        ],
    },
    {
        "decision_id": "DF04",
        "decision": "O campo governado nb14_score_source_recommendation determina a fonte primária dos escores no NB14/NB15.",
        "basis": "A consolidação decisória deve distinguir fonte primária NB11_FULL de sensibilidades NB13_FULL/NB13a_FULL.",
        "final_position": "NB11_FULL é a fonte primária nas oito células; NB13_FULL/NB13a_FULL entram apenas como sensibilidade nas células modeláveis.",
        "evidence_patterns": [
            "14_FULL_decision_analysis/aggregate/14_FULL_scenario_summary_by_cell.csv",
            "14_FULL_decision_analysis/aggregate/14_FULL_nb14_summary.json",
            "15_FULL_visualization/aggregate/figures/15_FULL_figD_score_source_by_cell.png",
        ],
    },
    {
        "decision_id": "DF05",
        "decision": "Calibração é diagnóstico de qualidade do escore, não transformação aplicada à política de decisão.",
        "basis": "Os escores são interpretados como risco temporal relativo; a calibração observada não autoriza leitura como probabilidade operacional plenamente calibrada.",
        "final_position": "Manter tau/custo sobre escore relativo governado; reportar calibração como limitação/diagnóstico.",
        "evidence_patterns": [
            "15_FULL_visualization/aggregate/15_FULL_calibration_diagnostics_by_cell.csv",
            "11_FULL_model_baselines/aggregate/11_FULL_calibration_summary_by_cell.csv",
            "14_FULL_decision_analysis/aggregate/14_FULL_scores_selected_allcells.parquet",
        ],
    },
    {
        "decision_id": "DF06",
        "decision": "O temporal_core permanece travado com 14 atributos para comparabilidade canônico ↔ _FULL.",
        "basis": "A replicação _FULL deve mudar população, célula e escala, não o núcleo de atributos temporais que sustenta a formulação.",
        "final_position": "Preservar temporal_core=14; atributos extras ou alternativos não são introduzidos no fechamento.",
        "evidence_patterns": [
            "05_FULL_features/aggregate/05_FULL_feature_sets.json",
            "11_FULL_model_baselines/aggregate/11_FULL_feature_sets.json",
            "11_FULL_model_baselines/aggregate/11_FULL_model_input_features_schema.json",
        ],
    },
    {
        "decision_id": "DF07",
        "decision": "As células Borg são réplicas independentes; os dados não devem ser concatenados como uma única série temporal.",
        "basis": "O ramo _FULL mede robustez/heterogeneidade por célula, preservando limiar, split, modelabilidade, métricas e decisão por unidade experimental.",
        "final_position": "Reportar por célula e consolidar apenas em tabelas/figuras agregadas interpretativas.",
        "evidence_patterns": [
            "04_FULL_episodes/aggregate/04_FULL_episode_summary_by_cell.csv",
            "11_FULL_model_baselines/aggregate/11_FULL_metrics_summary_by_cell.csv",
            "14_FULL_decision_analysis/aggregate/14_FULL_scenario_summary_by_cell.csv",
            "15_FULL_visualization/aggregate/15_FULL_visual_manifest_allcells.csv",
        ],
    },
    {
        "decision_id": "DF08",
        "decision": "O ramo _FULL é experimento ampliado de robustez/escala, não substituto do pipeline canônico.",
        "basis": "A estratégia _FULL testa se a formulação permanece bem-definida em população completa e heterogênea, sem reabrir a prova de conceito canônica.",
        "final_position": "Usar o _FULL como evidência complementar de robustez externa, escala e limites; manter o canônico como núcleo principal da defesa.",
        "evidence_patterns": [
            "04_FULL_episodes/aggregate/04_FULL_delta_vs_canonical.csv",
            "14_FULL_decision_analysis/aggregate/14_FULL_delta_vs_canonical.csv",
            "15_FULL_visualization/aggregate/15_FULL_delta_vs_canonical.csv",
        ],
    },
]

limitation_specs: List[Dict[str, Any]] = [
    {
        "limitation_id": "LF01",
        "limitation": "Heterogeneidade forte entre células e queda de desempenho em parte do domínio _FULL.",
        "basis": "A mudança de população para Borg completo por célula altera regime, prevalência e estabilidade temporal.",
        "final_position": "Reportar desempenho por célula e evitar média global como evidência única.",
        "evidence_patterns": [
            "11_FULL_model_baselines/aggregate/11_FULL_metrics_summary_by_cell.csv",
            "14_FULL_decision_analysis/aggregate/14_FULL_scenario_summary_by_cell.csv",
            "15_FULL_visualization/aggregate/figures/15_FULL_figA_dashboard_by_cell.png",
        ],
    },
    {
        "limitation_id": "LF02",
        "limitation": "Células de regime baixo, especialmente e/h, têm menor suporte empírico e maior sensibilidade à classe rara.",
        "basis": "O gate de modelabilidade e os diagnósticos descritivos por célula registram diferenças de suporte e prevalência.",
        "final_position": "Tratar essas células como evidência de limitação/robustez, não como falha do pipeline.",
        "evidence_patterns": [
            "04_FULL_episodes/aggregate/04_FULL_modelability_gate_by_cell.csv",
            "10_FULL_threshold_diagnostics/aggregate/10_FULL_fail_rate_descriptive_stats_by_cell.csv",
            "11_FULL_model_baselines/aggregate/11_FULL_modelability_by_cell_scenario.csv",
        ],
    },
    {
        "limitation_id": "LF03",
        "limitation": "Célula d não segue como célula modelável completa no ramo de LSTM/sensibilidade sequencial.",
        "basis": "Os artefatos de modelabilidade e tuning indicam suporte insuficiente/condição não modelável para a célula.",
        "final_position": "Manter cell_d no inventário e na descrição; não forçar modelagem sequencial onde o gate metodológico não sustenta.",
        "evidence_patterns": [
            "04_FULL_episodes/aggregate/04_FULL_modelability_gate_by_cell.csv",
            "11_FULL_model_baselines/aggregate/11_FULL_modelability_cell_overview.csv",
            "13_FULL_lstm/aggregate/13_FULL_lstm_skips_by_cell.csv",
            "13a_FULL_lstm_tuning/aggregate/13a_FULL_lstm_tuning_issues_allcells.csv",
        ],
    },
    {
        "limitation_id": "LF04",
        "limitation": "A LSTM não foi promovida a fonte primária; seu uso permanece complementar.",
        "basis": "Mesmo com tuning, o critério conservador Δ/SE ≥ 2,0 não justificou substituir NB11_FULL.",
        "final_position": "Discutir a LSTM como evidência de sensibilidade arquitetural, não como resultado principal.",
        "evidence_patterns": [
            "13_FULL_lstm/aggregate/13_FULL_lstm_delta_uncertainty_by_cell.csv",
            "13a_FULL_lstm_tuning/aggregate/13a_FULL_lstm_tuning_delta_vs_NB11_FULL_by_cell.csv",
            "13a_FULL_lstm_tuning/aggregate/13a_FULL_lstm_tuning_decision_to_nb14_by_cell.csv",
        ],
    },
    {
        "limitation_id": "LF05",
        "limitation": "A calibração dos escores é diagnóstica e não deve ser usada para afirmar probabilidade operacional calibrada.",
        "basis": "O ramo mantém interpretação de risco relativo e usa custo/tau de forma operacionalmente comparativa.",
        "final_position": "Evitar frases probabilísticas fortes; usar linguagem de ordenação relativa de risco.",
        "evidence_patterns": [
            "15_FULL_visualization/aggregate/15_FULL_calibration_diagnostics_by_cell.csv",
            "11_FULL_model_baselines/aggregate/11_FULL_calibration_summary_by_cell.csv",
            "14_FULL_decision_analysis/aggregate/14_FULL_threshold_metrics_by_tau_allcells.csv",
        ],
    },
    {
        "limitation_id": "LF06",
        "limitation": "NB13_FULL e NB13a_FULL são evidências complementares de sensibilidade sequencial, não fontes soberanas de decisão.",
        "basis": "As decisões nominais por ΔF1 e os diagnósticos de incerteza devem ser distinguidos da recomendação governante conservadora baseada em Δ/SE.",
        "final_position": "Reportar NB13_FULL/NB13a_FULL como análise complementar; preservar NB11_FULL como fonte primária quando a evidência sequencial não superar o gate conservador.",
        "evidence_patterns": [
            "13_FULL_lstm/aggregate/13_FULL_lstm_delta_uncertainty_by_cell.csv",
            "13_FULL_lstm/aggregate/13_FULL_lstm_decision_to_nb14_by_cell.csv",
            "13a_FULL_lstm_tuning/aggregate/13a_FULL_lstm_tuning_decision_to_nb14_by_cell.csv",
            "13a_FULL_lstm_tuning/aggregate/13a_FULL_lstm_tuning_delta_vs_NB11_FULL_by_cell.csv",
        ],
    },
    {
        "limitation_id": "LF07",
        "limitation": "O _FULL não substitui a prova de conceito canônica e não deve ser usado para reabrir decisões metodológicas seladas.",
        "basis": "A finalidade do ramo é robustez/escala, com delta mínimo e sem NB07/NB08/NB09_FULL.",
        "final_position": "Usar como capítulo/discussão complementar de robustez externa, escala, heterogeneidade e limites.",
        "evidence_patterns": [
            "10_FULL_threshold_diagnostics/aggregate/10_FULL_delta_vs_canonical.csv",
            "11_FULL_model_baselines/aggregate/11_FULL_delta_vs_canonical.csv",
            "15_FULL_visualization/aggregate/15_FULL_delta_vs_canonical.csv",
        ],
    },
]

decision_log_df = build_evidence_df(decision_specs)
# Remove coluna limitation vazia para manter estrutura canônica do decision_log.
decision_log_df = decision_log_df[
    ["decision_id", "decision", "basis", "final_position", "evidence_artifacts", "evidence_status", "evidence_found", "evidence_missing"]
].copy()

limitations_df = build_evidence_df(limitation_specs)
limitations_df = limitations_df[
    ["limitation_id", "limitation", "basis", "final_position", "evidence_artifacts", "evidence_status", "evidence_found", "evidence_missing"]
].copy()

# Checks explícitos para que a ausência de evidência em decisão/limitação apareça no
# fechamento operacional do NB16_FULL.
decision_status_counts = decision_log_df["evidence_status"].value_counts(dropna=False).to_dict()
limitation_status_counts = limitations_df["evidence_status"].value_counts(dropna=False).to_dict()

add_check(
    consistency_rows,
    "Decision log estruturado DF01–DF08",
    "OK" if set(decision_log_df["evidence_status"]) <= {"existe"} else "WARN",
    f"evidence_status={decision_status_counts}",
    "warning" if set(decision_log_df["evidence_status"]) - {"existe"} else "info",
)
add_check(
    consistency_rows,
    "Registro estruturado de limitações LF01–LF07",
    "OK" if set(limitations_df["evidence_status"]) <= {"existe"} else "WARN",
    f"evidence_status={limitation_status_counts}",
    "warning" if set(limitations_df["evidence_status"]) - {"existe"} else "info",
)

# Recria o DataFrame de checks após os dois registros estruturados.
consistency_df = pd.DataFrame(consistency_rows)

# -----------------------------------------------------------------------------
# 8. Delta canônico ↔ FULL e registro de figuras da dissertação
# -----------------------------------------------------------------------------

delta_rows = [
    {"notebook_full": "16_final_inventory_FULL.ipynb", "notebook_canonical": "16_final_inventory.ipynb", "change_type": "I/O_PATH", "description": "Troca de 04-reports canônico por 04-reports/99_FULL_downstream.", "justified": True},
    {"notebook_full": "16_final_inventory_FULL.ipynb", "notebook_canonical": "16_final_inventory.ipynb", "change_type": "OUTPUT_PREFIX", "description": "Todos os artefatos produzidos usam prefixo 16_FULL_.", "justified": True},
    {"notebook_full": "16_final_inventory_FULL.ipynb", "notebook_canonical": "16_final_inventory.ipynb", "change_type": "AGGREGATION_BY_CELL", "description": "Inclusão da cobertura por célula a-h e da matriz de rastreabilidade por stage/cell.", "justified": True},
    {"notebook_full": "16_final_inventory_FULL.ipynb", "notebook_canonical": "16_final_inventory.ipynb", "change_type": "FULL_DATASET_INPUT", "description": "Registro de proveniência do Parquet model-facing FULL selado pelo NB99_C.", "justified": True},
    {"notebook_full": "16_final_inventory_FULL.ipynb", "notebook_canonical": "16_final_inventory.ipynb", "change_type": "REPORTING_ONLY", "description": "NB07_FULL/NB08_FULL/NB09_FULL são explicitamente não esperados no ramo FULL.", "justified": True},
    {"notebook_full": "16_final_inventory_FULL.ipynb", "notebook_canonical": "16_final_inventory.ipynb", "change_type": "REPORTING_ONLY", "description": "NB13a_FULL é tratado como complementar/opcional.", "justified": True},
    {"notebook_full": "16_final_inventory_FULL.ipynb", "notebook_canonical": "16_final_inventory.ipynb", "change_type": "REPORTING_ONLY", "description": "Síntese narrativa enfatiza robustez/escala e não substituição do canônico.", "justified": True},
]
delta_vs_canonical_df = pd.DataFrame(delta_rows)

# Registro de figuras FULL a partir do manifesto visual ou varredura de arquivos.
fig_rows: List[Dict[str, Any]] = []
if df15_manifest is not None and not df15_manifest.empty:
    filename_col = first_col(df15_manifest, ["filename", "file", "figure_file"])
    path_col = first_col(df15_manifest, ["path", "absolute_path", "relative_path"])
    cell_col = first_col(df15_manifest, ["cell_id", "cell"])
    fig_col = first_col(df15_manifest, ["figure_type", "fig_type"])
    for _, r in df15_manifest.iterrows():
        filename = str(r.get(filename_col, "")) if filename_col else ""
        fig_type = str(r.get(fig_col, "")) if fig_col else ""
        cid = str(r.get(cell_col, "")) if cell_col else ""
        raw_path = str(r.get(path_col, "")) if path_col else ""
        citable = "appendix"
        if cid == "aggregate" or "dashboard_by_cell" in fig_type or filename.startswith("15_FULL_figA"):
            citable = "yes"
        elif "temporal_zoom" in fig_type or "temporal_overview" in fig_type:
            citable = "appendix"
        fig_rows.append({
            "source_notebook": "15_FULL",
            "cell_id": cid,
            "figure_type": fig_type,
            "filename": filename,
            "path": raw_path,
            "recommended_use": citable,
            "role": "Figura gerada pelo NB15_FULL; uso principal em robustez/escala ou apêndice por célula.",
        })
else:
    fig_dir = FULL_REPORTS_DIR / "15_FULL_visualization"
    if fig_dir.exists():
        for p in sorted(fig_dir.rglob("15_FULL_fig*")):
            if p.is_file() and p.suffix.lower() in {".png", ".pdf", ".svg", ".jpg", ".jpeg"}:
                cid = infer_cell_from_path(p) or ("aggregate" if "aggregate" in p.parts else "")
                fig_rows.append({
                    "source_notebook": "15_FULL",
                    "cell_id": cid,
                    "figure_type": "inferred_from_filename",
                    "filename": p.name,
                    "path": relpath(p),
                    "recommended_use": "yes" if cid == "aggregate" else "appendix",
                    "role": "Figura encontrada por varredura do diretório NB15_FULL.",
                })

figures_registry_df = pd.DataFrame(fig_rows)

# -----------------------------------------------------------------------------
# 9. Síntese narrativa e notas de fechamento
# -----------------------------------------------------------------------------

def md_table_from_df(df: pd.DataFrame, cols: Sequence[str], max_rows: int = 30) -> str:
    if df is None or df.empty:
        return "_Sem dados disponíveis._"
    sub = df.loc[:, [c for c in cols if c in df.columns]].head(max_rows).copy()
    return sub.to_markdown(index=False)

status_counts = traceability_df["status"].value_counts(dropna=False).to_dict() if not traceability_df.empty else {}
check_counts = consistency_df["status"].value_counts(dropna=False).to_dict() if not consistency_df.empty else {}
warning_total = int((consistency_df["status"].isin(["WARN", "FAIL"])).sum()) if not consistency_df.empty else 0
fail_total = int((consistency_df["status"] == "FAIL").sum()) if not consistency_df.empty else 0
inventory_total = int(len(inventory_df))
manifest_total = int(len(manifest_df))
figures_total = int(len(figures_registry_df))

trace_md = md_table_from_df(traceability_df, ["nb", "stage_dir", "artifact_count", "cells_with_artifacts_count", "status", "notes"], max_rows=20)
checks_md = md_table_from_df(consistency_df, ["check", "status", "severity", "evidence"], max_rows=40)
key_md = md_table_from_df(key_results_df, ["stage", "source", "indicator", "value", "interpretation"], max_rows=40)
decision_md = md_table_from_df(decision_log_df, ["decision_id", "decision", "final_position", "evidence_status"], max_rows=20)
limitations_md = md_table_from_df(limitations_df, ["limitation_id", "limitation", "final_position", "evidence_status"], max_rows=20)

narrative_synthesis_md = f"""# NB16_FULL — Síntese narrativa do fechamento do ramo `_FULL`

_Gerado automaticamente em {EXECUTION_TS}._

## 1. Papel do fechamento

O `NB16_FULL` fecha documentalmente o ramo `_FULL` do pipeline PPCOMP/DM. A etapa não recalcula nenhum resultado experimental; ela consolida inventário, rastreabilidade, cobertura por célula, checks de consistência e síntese narrativa a partir dos artefatos já produzidos.

## 2. Enquadramento metodológico

O ramo `_FULL` deve ser interpretado como replicação ampliada de robustez e escala sobre o Borg 2019 completo por célula. Ele não substitui o pipeline canônico. A célula Borg permanece a unidade experimental, e os agregados têm função de síntese comparativa.

## 3. Resultado de rastreabilidade

Resumo de status das etapas: `{status_counts}`.

{trace_md}

## 4. Inventário e manifesto

Foram inventariados `{inventory_total}` artefatos, com `{manifest_total}` registros de manifesto SHA-256. Arquivos acima de `{HASH_MAX_BYTES}` bytes podem ter hash omitido com status `skipped_gt_*`, preservando a rastreabilidade sem custo excessivo.

## 5. Cobertura visual e figuras

Foram registrados `{figures_total}` itens no registro de figuras `_FULL`. Figuras agregadas do NB15_FULL são candidatas a uso no corpo ou na discussão de robustez; figuras por célula tendem a ser mais adequadas para apêndice ou análise complementar.

## 6. Resultados-chave lidos dos artefatos

{key_md}

## 7. Checks de consistência

Resumo de checks: `{check_counts}`.

{checks_md}

## 8. Registro estruturado de decisões

O fechamento inclui `16_FULL_decision_log.csv`, espelhando o papel do decision log canônico, com decisões DF01–DF08 e verificação automática de evidências.

{decision_md}

## 9. Registro estruturado de limitações

O fechamento inclui `16_FULL_limitations.csv`, com limitações LF01–LF07 e verificação automática de evidências.

{limitations_md}

## 10. Decisão de fechamento

Se não houver `FAIL` crítico, o ramo `_FULL` pode ser considerado fechado documentalmente para fins de dissertação. Se houver `WARN`, a etapa correspondente deve ser revisada manualmente, sem que o NB16_FULL tente alterar ou corrigir os artefatos de origem.

## 11. Uso recomendado na dissertação

Os artefatos do `NB16_FULL` devem apoiar a discussão de robustez externa, heterogeneidade entre células, rastreabilidade dos notebooks e delimitação entre prova de conceito canônica e replicação ampliada no Borg completo.
"""

closure_notes_md = f"""# NB16_FULL — Notas automáticas de conclusão da etapa

## 1. Resultado operacional

| Item | Resultado |
|---|---:|
| Versão do notebook | {NB16_FULL_VERSION} |
| Timestamp | {EXECUTION_TS} |
| Artefatos inventariados | {inventory_total} |
| Registros no manifesto | {manifest_total} |
| Figuras registradas | {figures_total} |
| Checks com WARN/FAIL | {warning_total} |
| FAIL críticos | {fail_total} |

## 2. Status por etapa

{trace_md}

## 3. Decisões e limitações estruturadas

| Registro | Linhas | Status de evidência |
|---|---:|---|
| `16_FULL_decision_log.csv` | {len(decision_log_df)} | {decision_status_counts} |
| `16_FULL_limitations.csv` | {len(limitations_df)} | {limitation_status_counts} |

## 4. Interpretação

- `OK`: etapa localizada e suficientemente rastreável.
- `WARN`: etapa localizada parcialmente, com ausência de padrão agregado ou cobertura incompleta por célula.
- `FAIL`: etapa obrigatória não localizada.
- `SKIP_OPTIONAL`: etapa complementar não executada, sem bloqueio metodológico.

## 5. Decisão sugerida

O ramo `_FULL` está documentalmente fechado se `FAIL críticos = 0`. Warnings devem ser lidos como pontos de revisão de diretório, nomenclatura ou cobertura por célula, não como autorização para recálculo nesta etapa.
"""

# -----------------------------------------------------------------------------
# 10. Persistência dos artefatos NB16_FULL
# -----------------------------------------------------------------------------

outputs: Dict[str, str] = {}

for filename, df in [
    ("16_FULL_artifact_inventory.csv", inventory_df),
    ("16_FULL_artifact_manifest_sha256.csv", manifest_df),
    ("16_FULL_traceability_matrix.csv", traceability_df),
    ("16_FULL_expected_artifacts_check.csv", expected_df),
    ("16_FULL_cell_stage_coverage.csv", coverage_df),
    ("16_FULL_key_results_summary.csv", key_results_df),
    ("16_FULL_consistency_checks.csv", consistency_df),
    ("16_FULL_decision_log.csv", decision_log_df),
    ("16_FULL_limitations.csv", limitations_df),
    ("16_FULL_delta_vs_canonical.csv", delta_vs_canonical_df),
    ("16_FULL_dissertation_figures_registry.csv", figures_registry_df),
]:
    p = save_csv(df, filename)
    outputs[filename] = str(p)

for filename, text in [
    ("16_FULL_narrative_synthesis.md", narrative_synthesis_md),
    ("16_FULL_execution_closure_notes.md", closure_notes_md),
]:
    p = save_md(text, filename)
    outputs[filename] = str(p)

# Summary executivo. Neste ponto, o inventário principal já representa somente
# artefatos-fonte do ramo _FULL, excluindo os próprios outputs do NB16_FULL.
summary = {
    "notebook": "NB16_FULL",
    "version": NB16_FULL_VERSION,
    "title": "Consolidação, inventário, rastreabilidade e fechamento do ramo _FULL",
    "execution_ts": EXECUTION_TS,
    "project_root": str(PROJECT_ROOT),
    "full_reports_dir": str(FULL_REPORTS_DIR),
    "stage16_dir": str(STAGE16_DIR),
    "aggregate_dir": str(AGGREGATE_DIR),
    "full_model_facing_parquet": str(FULL_MODEL_FACING_PARQUET),
    "full_model_facing_exists": bool(FULL_MODEL_FACING_PARQUET.exists()),
    "inventory_scope": {
        "source_inventory_excludes_nb16_outputs": bool(EXCLUDE_NB16_OUTPUTS_FROM_SOURCE_INVENTORY),
        "source_inventory_description": "Artefatos-fonte do downstream _FULL, principalmente NB04_FULL–NB15_FULL, mais proveniência model-facing quando localizada.",
        "nb16_outputs_manifest": "16_FULL_outputs_manifest_sha256.csv",
        "nb16_residuals_removed_from_source_inventory": int(nb16_residuals_removed),
    },
    "central_rule": {
        "does_not_recalculate": True,
        "does_not_change_data": True,
        "does_not_open_new_experiments": True,
        "purpose": "inventariar, validar, consolidar e documentar o downstream _FULL",
    },
    "cells": {
        "full_cells": FULL_CELLS,
        "active_cells": ACTIVE_CELLS,
    },
    "scenario_main": SCENARIO_MAIN,
    "canonical_temporal_core": CANONICAL_TEMPORAL_CORE,
    "artifact_counts": {
        "source_inventory_total": inventory_total,
        "source_manifest_total": manifest_total,
        "nb16_residuals_removed_from_source_inventory": int(nb16_residuals_removed),
        "figures_registry_total": figures_total,
        "key_results_total": int(len(key_results_df)),
        "consistency_checks_total": int(len(consistency_df)),
        "decision_log_total": int(len(decision_log_df)),
        "limitations_total": int(len(limitations_df)),
        "decision_log_evidence_status_counts": decision_status_counts,
        "limitations_evidence_status_counts": limitation_status_counts,
        "warnings_or_critical_total": warning_total,
        "fail_total": fail_total,
    },
    "traceability_status_counts": status_counts,
    "consistency_status_counts": check_counts,
}

summary_path = save_json(summary, "16_FULL_nb16_summary.json")
outputs["16_FULL_nb16_summary.json"] = str(summary_path)

# Manifesto dos próprios outputs gerados nesta execução.
# Este arquivo é separado do inventário-fonte para evitar autorreferência.
# O manifesto de outputs não contém hash de si mesmo, evitando circularidade.
output_manifest_rows = []
for filename, path_str in sorted(outputs.items()):
    p = Path(path_str)
    h, hs = sha256_file(p)
    output_manifest_rows.append({
        "filename": filename,
        "relative_path": relpath(p),
        "absolute_path": str(p),
        "size_bytes": p.stat().st_size if p.exists() else None,
        "sha256": h,
        "sha256_status": hs,
    })

outputs_manifest_df = pd.DataFrame(output_manifest_rows)
outputs_manifest_path = save_csv(outputs_manifest_df, "16_FULL_outputs_manifest_sha256.csv")
outputs["16_FULL_outputs_manifest_sha256.csv"] = str(outputs_manifest_path)

# Check final de resíduos NB16 no inventário-fonte.
if not inventory_df.empty:
    _residual_after_filter = inventory_df.apply(is_nb16_residual_row, axis=1)
    assert int(_residual_after_filter.sum()) == 0, "Ainda há resíduos NB16 no inventário-fonte."

# Check final dos prefixos gerados.
bad_output_names = [name for name in outputs if not name.startswith("16_FULL_")]
assert not bad_output_names, f"Saídas sem prefixo 16_FULL_: {bad_output_names}"

print("\n" + "=" * 100)
print("Artefatos gerados pelo NB16_FULL")
print("=" * 100)
for filename, path_str in outputs.items():
    print(f"- {filename}: {path_str}")

print("\nResumo final:")
print(json.dumps({
    "inventory_total": inventory_total,
    "manifest_total": manifest_total,
    "traceability_status_counts": status_counts,
    "consistency_status_counts": check_counts,
    "decision_log_evidence_status_counts": decision_status_counts,
    "limitations_evidence_status_counts": limitation_status_counts,
    "warnings_or_critical_total": warning_total,
    "fail_total": fail_total,
}, ensure_ascii=False, indent=2))

print("\nConclusão:")
if fail_total == 0:
    print("NB16_FULL concluído sem FAIL crítico. O ramo _FULL está documentalmente fechado, sujeito à leitura dos WARNs.")
else:
    print("NB16_FULL concluído com FAIL crítico. Revisar as etapas obrigatórias ausentes antes de usar o fechamento na dissertação.")
print("=" * 100)
````
